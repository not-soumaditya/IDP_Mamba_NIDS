# 📊 UNSW-NB15 Data Exploration

Exploratory data analysis of the UNSW-NB15 network intrusion detection dataset.

**Dataset:** UNSW-NB15 (University of New South Wales, 2015)  
**Task:** Classify network flows into Normal or 9 attack categories

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)

# Load raw data
train_df = pd.read_csv('../data/raw/UNSW_NB15_training-set.csv')
test_df = pd.read_csv('../data/raw/UNSW_NB15_testing-set.csv')

print(f'Train shape: {train_df.shape}')
print(f'Test shape:  {test_df.shape}')
train_df.head()
```

```python
# Basic info
print('Column types:')
print(train_df.dtypes.value_counts())
print(f'\nMissing values: {train_df.isnull().sum().sum()}')
train_df.describe()
```

```python
# Class distribution
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

train_df['attack_cat'].value_counts().plot(kind='barh', ax=axes[0], color='steelblue')
axes[0].set_title('Train Set — Attack Categories')
axes[0].set_xlabel('Count')

test_df['attack_cat'].value_counts().plot(kind='barh', ax=axes[1], color='coral')
axes[1].set_title('Test Set — Attack Categories')
axes[1].set_xlabel('Count')

plt.tight_layout()
plt.show()
```

```python
# Feature correlations (top numeric features)
numeric_cols = train_df.select_dtypes(include=[np.number]).columns[:20]
fig, ax = plt.subplots(figsize=(14, 12))
sns.heatmap(train_df[numeric_cols].corr(), cmap='coolwarm', center=0,
            annot=False, ax=ax)
ax.set_title('Feature Correlation Matrix (first 20 numeric features)')
plt.tight_layout()
plt.show()
```

```python
# Protocol distribution
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

train_df['proto'].value_counts().head(10).plot(kind='bar', ax=axes[0])
axes[0].set_title('Top 10 Protocols')
axes[0].tick_params(axis='x', rotation=45)

train_df['service'].value_counts().head(10).plot(kind='bar', ax=axes[1])
axes[1].set_title('Top 10 Services')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```
